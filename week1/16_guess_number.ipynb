```python
import random

answer = random.randint(1, 10)
print("随机答案是：", answer)
```

```python
guess = int(input("请输入你猜的数字(1-10):"))
if guess == answer:
    print("猜对了！")
elif guess < answer:
    print("太小了！")
else:
    print("太大了")
```

```python
while True:
    guess = int(input("请输入你猜的数字(1-10):"))

    if guess == answer:
        print("猜对了！")
        break
    elif guess < answer:
        print("太小了！")
    else:
        print("太大了！")
```

```python
attempts = 0
while True:
    guess = int(input("请输入你猜的数字(1-10):"))
    attempts = attempts + 1

    if guess == answer:
        print("猜对了！")
        print("你一共猜了", attempts, "次")
        break
    elif guess < answer:
        print("太小了！")
    else:
        print("太大了！")
```

```python
import random

answer = random.randint(1,10)
max_attempts = 5
attempts = 0

while attempts < max_attempts:
    guess = int(input("请输入你猜的数字(1-10):"))
    attempts = attempts + 1

    if guess == answer:
        print("猜对了！")
        print("你一共猜了",attempts,"次")
        break
    elif guess < answer:
        print("太小了！")
    else:
        print("太大了！")

if guess != answer:
    print("5次机会已经用完了！")
    print("正确答案是：",answer)
```
